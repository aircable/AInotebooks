```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("roneneldan/TinyStories-1M")
#tokenizer = AutoTokenizer.from_pretrained("EleutherAI/gpt-neo-125M")
#model = AutoModelForCausalLM.from_pretrained("roneneldan/TinyStories-1M")
model = AutoModelForCausalLM.from_pretrained('roneneldan/TinyStories-Instruct-1M')

print( tokenizer )
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
model = AutoModelForCausalLM.from_pretrained("roneneldan/TinyStories-Instruct-1M")
tok   = AutoTokenizer.from_pretrained("roneneldan/TinyStories-Instruct-1M")
print(model.config)      # shows hidden_size=64, num_heads=16, num_layers=8, vocab_size=50257
print("Parameters:", sum(p.numel() for p in model.parameters()))  # ~3.745e6
print("Vocab size:", tok.vocab_size)  # 50257
```

```python
prompt = "Once upon a time there was"

input_ids = tokenizer.encode(prompt, return_tensors="pt")
```

```python
output = model.generate(input_ids, max_length = 1000, num_beams=1)
```

```python
output_text = tokenizer.decode(output[0], skip_special_tokens=True)
```

```python
print(output_text)
```
